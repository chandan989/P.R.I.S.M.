```python
from unsloth import FastModel
import torch
from google.colab import userdata
import os

# Implement ModelScope workaround for HuggingFace timeout
os.environ['UNSLOTH_USE_MODELSCOPE'] = '1'

# 1. Correct the name (added 26b-)
model_name = "google/gemma-4-26b-a4b-it"

# 2. You MUST provide a token for gated Gemma models
# You can set this in the Colab 'Secrets' tab (🔑) as HF_TOKEN
hf_token = userdata.get("HF_TOKEN")

model, tokenizer = FastModel.from_pretrained(
    model_name = model_name,
    max_seq_length = 8192,
    load_in_4bit = True,
    token = hf_token, # Required for gated models
)
```

```python
print("Setting up PEFT / LoRA Adapters...")
model = FastModel.get_peft_model(
    model,
    r = 32,
    target_modules = [
        "q_proj", "k_proj", "v_proj", "o_proj",
        "gate_proj", "up_proj", "down_proj", # Essential for MoE Experts
        "moe_gate", # Critical for Gemma 4 MoE Router
        "embed_tokens", "lm_head" # CRITICAL: Allows learning new tokens
    ],
    lora_alpha = 16,
    lora_dropout = 0,
    use_dora = False,
    bias = "none",
)
```

```python
from datasets import Dataset

print("Preparing Polypharmacy Deliberation Dataset...")

system_prompt = """<system>
For every response:
1. Begin reasoning with <|think|> to expose your deliberation
2. Enumerate competing interpretations with probability estimates
3. Use tool calls to verify factual claims when confidence is low
4. Provide a calibrated confidence score for each major conclusion
</system>"""

prompt_template = system_prompt + """
<bos><start_of_turn>user
{instruction}<end_of_turn>
<start_of_turn>model
<|think|>
<|channel>thought
{thought_process}
<|channel>
{output}<end_of_turn><eos>"""

def format_prompts(examples):
    instructions = examples["instruction"]
    thoughts = examples["thought_process"]
    outputs = examples["output"]
    texts = []
    for instruction, thought, output in zip(instructions, thoughts, outputs):
        text = prompt_template.format(instruction=instruction, thought_process=thought, output=output)
        texts.append(text)
    return { "text" : texts }

import json
import os

print("Preparing Polypharmacy Deliberation Dataset...")

# Use the project_dir defined in your Drive mount cell
dataset_path = os.path.join(project_dir, "deliberation_dataset.json")

if not os.path.exists(dataset_path):
    raise FileNotFoundError(f"Dataset missing at {dataset_path}. Please ensure it is uploaded.")

with open(dataset_path, "r") as f:
    data_list = json.load(f)

# Convert to dictionary of lists for Hugging Face Dataset format
data = {
    "instruction": [item["instruction"] for item in data_list],
    "thought_process": [item["thought_process"] for item in data_list],
    "output": [item["output"] for item in data_list]
}

deliberation_dataset = Dataset.from_dict(data)
deliberation_dataset = deliberation_dataset.map(format_prompts, batched=True)
```

### Base Model Evaluation (Before Training)
Let's test the model's formatting and output on a holdout polypharmacy prompt before it learns the P.R.I.S.M. deliberation structure.

```python
FastModel.for_inference(model)

eval_prompt = "Patient on Warfarin 5mg and Omeprazole 20mg. What are the interactions?"
prompt = f"{system_prompt}\n<bos><start_of_turn>user\n{eval_prompt}<end_of_turn>\n<start_of_turn>model\n<|think|>\n"

# Use explicit 'text=' keyword to bypass the Gemma 4 multimodal Processor TypeError
inputs = tokenizer(
    text = [prompt],
    return_tensors = "pt",
    add_special_tokens = False
).to("cuda")

# Increased max_new_tokens to allow full structured deliberation
outputs = model.generate(**inputs, max_new_tokens = 1024, use_cache = True)
base_response = tokenizer.batch_decode(outputs, skip_special_tokens = True)[0]

print("========== BASE MODEL RESPONSE ==========")
print(base_response)
```

```python
from trl import SFTTrainer, SFTConfig # <-- CRITICAL: Import SFTConfig instead of TrainingArguments
import os
import torch

print("Starting SFT Training...")
is_bfloat16 = torch.cuda.is_bf16_supported()

trainer = SFTTrainer(
    model = model,
    processing_class = tokenizer.tokenizer,
    train_dataset = deliberation_dataset,
    dataset_text_field = "text",
    max_seq_length = 8192,
    # CRITICAL FIX: Use SFTConfig directly to bypass Unsloth's broken translation layer
    args = SFTConfig(
        per_device_train_batch_size = 4,
        gradient_accumulation_steps = 2,
        warmup_steps = 5,
        num_train_epochs = 15, # Replaced max_steps with num_train_epochs to ensure enough training for new tokens
        learning_rate = 2e-4,
        fp16 = not is_bfloat16,
        bf16 = is_bfloat16,
        logging_steps = 1,
        optim = "adamw_8bit",
        weight_decay = 0.01,
        lr_scheduler_type = "linear",
        seed = 3407,
        output_dir = os.path.join(project_dir, "outputs"),
        dataset_text_field = "text", # Moved inside SFTConfig as required by newer TRL
        max_seq_length = 8192,       # Moved inside SFTConfig as required by newer TRL
    ),
)

trainer.train()
```

### Fine-Tuned Model Evaluation (After Training)
Now let's run the exact same prompt through the model to see if it successfully adopted the structured `<|think|>` deliberation format and source verification.

```python
FastModel.for_inference(model) # Enable native 2x faster inference

# 1. Use explicit 'text=' keyword to bypass the 'NoneType' image bug
# 2. Prevent double-BOS tokens if your prompt string already has one
inputs = tokenizer(
    text = [prompt],
    return_tensors = "pt",
    add_special_tokens = False
).to("cuda")

# Increased to 1024 to accommodate complex hypothesis enumeration
outputs = model.generate(**inputs, max_new_tokens = 1024, use_cache = True)
# CRITICAL FIX: skip_special_tokens must be False to see your custom tokens!
tuned_response = tokenizer.batch_decode(outputs, skip_special_tokens = False)[0]

print("========== FINE-TUNED MODEL RESPONSE ==========")
print(tuned_response)
print("=========================================")

def evaluate_structure(response):
    metrics = {
        "Has Logical Chain": "[Logical Chain]" in response,
        "Has Competing Hypotheses": "[Competing Hypotheses]" in response,
        "Has Discarded Paths": "✗ Discarded:" in response or "[Discarded Paths]" in response,
        "Has Tool Call": "<|tool_call>verify_claim" in response and "<|\"|>" in response,
        "Has Final Output Signal": "🟢" in response or "🟡" in response or "🔴" in response,
        "Has Confidence Badge": "Confidence:" in response
    }
    print("\n--- Structured Output Evaluation ---")
    score = sum(1 for passed in metrics.values() if passed)
    for key, passed in metrics.items():
        print(f"{key}: {'✅ Pass' if passed else '❌ Fail'}")
    print(f"\nOverall Structure Score: {score}/{len(metrics)}")

print("\nEvaluating Base Model:")
evaluate_structure(base_response)

print("\nEvaluating Fine-Tuned Model:")
evaluate_structure(tuned_response)
```

```python
!pip install -q kaggle

import os
import json
from huggingface_hub import hf_hub_download
from google.colab import userdata

# 1. Setup Kaggle API credentials from Colab Secrets
try:
    os.environ['KAGGLE_USERNAME'] = userdata.get('KAGGLE_USERNAME')
    os.environ['KAGGLE_KEY'] = userdata.get('KAGGLE_KEY')
except Exception as e:
    print("Error: KAGGLE_USERNAME and KAGGLE_KEY must be set in Colab Secrets.")
    raise e

import kaggle

# 2. Download from Hugging Face (will use cache if already downloaded/uploaded in the same session)
print("Downloading model from Hugging Face (or retrieving from cache)...")
hf_repo_id = "chandan989/prism-gemma-4-26B-A4B-it-MXFP4_MOE-v1"
hf_filename = "prism-gemma-4-26B-A4B-it-MXFP4_MOE-v1.gguf"

local_hf_path = hf_hub_download(
    repo_id=hf_repo_id,
    filename=hf_filename,
    token=userdata.get('HF_TOKEN')
)

# 3. Prepare Kaggle Dataset Directory
kaggle_dir = "kaggle_upload_dir"
os.makedirs(kaggle_dir, exist_ok=True)

# Symlink the downloaded file into the Kaggle directory to save space
dest_path = os.path.join(kaggle_dir, hf_filename)
if not os.path.exists(dest_path):
    os.symlink(local_hf_path, dest_path)

# 4. Create dataset-metadata.json
kaggle_username = os.environ['KAGGLE_USERNAME']
dataset_slug = "prism-gemma-4-26b-a4b-it-mxfp4-v1" # Must be lowercase, alphanumeric, and hyphens only
dataset_title = "PRISM Gemma-4-26B-A4B-IT MXFP4"

metadata = {
    "title": dataset_title,
    "id": f"{kaggle_username}/{dataset_slug}",
    "licenses": [{"name": "other"}]
}

with open(os.path.join(kaggle_dir, "dataset-metadata.json"), "w") as f:
    json.dump(metadata, f, indent=4)

# 5. Upload to Kaggle
print(f"Uploading {dataset_title} to Kaggle...")
try:
    kaggle.api.dataset_create_new(
        folder=kaggle_dir,
        dir_mode='skip' # 'skip' avoids zipping the file, uploading it as a raw GGUF
    )
    print(f"Upload complete! Check your Kaggle account: https://www.kaggle.com/{kaggle_username}/{dataset_slug}")
except Exception as e:
    if "already exists" in str(e).lower():
        print("Dataset already exists. Updating instead...")
        kaggle.api.dataset_create_version(
            folder=kaggle_dir,
            version_notes="Initial MXFP4 release"
        )
        print(f"Update complete! Check your Kaggle account: https://www.kaggle.com/{kaggle_username}/{dataset_slug}")
    else:
        raise e
```

### 🗺️ P.R.I.S.M. Token Map

The following table defines the mapping between human-readable deliberation tags and the model's special `<unusedX>` tokens:

| Category | Human-Readable Token | Model Token | Description |
| :--- | :--- | :--- | :--- |
| 🏷️ **Tags** | `<|think|>` | `<unused0>` | Initiates the deliberation phase |
| 🏷️ **Tags** | `<|channel>thought` | `<unused1>` | Opens the internal thought channel |
| 🏷️ **Tags** | `<|channel>` | `<unused2>` | Closes a specific channel |
| 🏷️ **Tags** | `<|tool_call>` | `<unused3>` | Triggers external tool invocation |
| 🏷️ **Tags** | `<|"|>` | `<unused4>` | Parameter boundary for tool calls |
| 📑 **Headers** | `[Logical Chain]` | `<unused5>` | Step-by-step logical deduction |
| 📑 **Headers** | `[Competing Hypotheses]` | `<unused6>` | Enumeration of possible interpretations |
| 📑 **Headers** | `[Discarded Paths]` | `<unused7>` | Explicit rejection of incorrect hypotheses |
| 📑 **Headers** | `▶ Selected:` | `<unused8>` | Final conclusion selection |
| 📑 **Headers** | `✗ Discarded:` | `<unused9>` | Explicit rejection label |

```python
from datasets import Dataset

print("Preparing Polypharmacy Deliberation Dataset...")

system_prompt = """For every response:
1. Begin reasoning with <unused0> to expose your deliberation
2. Enumerate competing interpretations with probability estimates
3. Use tool calls to verify factual claims when confidence is low
4. Provide a calibrated confidence score for each major conclusion"""

prompt_template = system_prompt + """
<bos><start_of_turn>user
{instruction}<end_of_turn>
<start_of_turn>model
<unused0>
<unused1>
{thought_process}
<unused2>
{output}<end_of_turn><eos>"""

def format_prompts(examples):
    instructions = examples["instruction"]
    thoughts = examples["thought_process"]
    outputs = examples["output"]
    texts = []

    for instruction, thought, output in zip(instructions, thoughts, outputs):
        # Map headers in thought process
        thought = thought.replace("[Logical Chain]", "<unused5>")
        thought = thought.replace("[Competing Hypotheses]", "<unused6>")
        thought = thought.replace("[Discarded Paths]", "<unused7>")
        thought = thought.replace("▶ Selected:", "<unused8>")
        thought = thought.replace("✗ Discarded:", "<unused9>")

        # CRITICAL FIX: Map tool calls inside the thought process, not output
        thought = thought.replace("<|tool_call>", "<unused3>")
        thought = thought.replace('<|"|>', "<unused4>")

        text = prompt_template.format(instruction=instruction, thought_process=thought, output=output)
        texts.append(text)

    return { "text": texts }

import json
import os

print("Preparing Polypharmacy Deliberation Dataset...")

# Use the project_dir defined in your Drive mount cell
dataset_path = os.path.join(project_dir, "deliberation_dataset.json")

if not os.path.exists(dataset_path):
    raise FileNotFoundError(f"Dataset missing at {dataset_path}. Please ensure it is uploaded.")

with open(dataset_path, "r") as f:
    data_list = json.load(f)

# Convert to dictionary of lists for Hugging Face Dataset format
data = {
    "instruction": [item["instruction"] for item in data_list],
    "thought_process": [item["thought_process"] for item in data_list],
    "output": [item["output"] for item in data_list]
}

deliberation_dataset = Dataset.from_dict(data)
deliberation_dataset = deliberation_dataset.map(format_prompts, batched=True)
```

### Base Model Evaluation (Before Training)
Let's test the model's formatting and output on a holdout polypharmacy prompt before it learns the P.R.I.S.M. deliberation structure.

---
**1st Iteration Hyperparameters (For Record):**
- `lora_alpha`: 16
- `learning_rate`: 2e-4
- `warmup_steps`: 5
- `num_train_epochs`: 15
- `lr_scheduler_type`: linear
- `special_tokens_registered`: False

```python
FastModel.for_inference(model)

eval_prompt = "Patient on Warfarin 5mg and Omeprazole 20mg. What are the interactions?"
prompt = f"{system_prompt}\n<bos><start_of_turn>user\n{eval_prompt}<end_of_turn>\n<start_of_turn>model\n<unused0>\n"

# Use explicit 'text=' keyword to bypass the Gemma 4 multimodal Processor TypeError
inputs = tokenizer(
    text = [prompt],
    return_tensors = "pt",
    add_special_tokens = False
).to("cuda")

# Increased max_new_tokens to allow full structured deliberation
outputs = model.generate(**inputs, max_new_tokens = 1024, use_cache = True)
base_response = tokenizer.batch_decode(outputs, skip_special_tokens = True)[0]

print("========== BASE MODEL RESPONSE ==========")
print(base_response)
```

```python
from trl import SFTTrainer, SFTConfig # <-- CRITICAL: Import SFTConfig instead of TrainingArguments
import os
import torch

print("Starting SFT Training...")
is_bfloat16 = torch.cuda.is_bf16_supported()

trainer = SFTTrainer(
    model = model,
    processing_class = tokenizer.tokenizer,
    train_dataset = deliberation_dataset,
    dataset_text_field = "text",
    max_seq_length = 8192,
    # CRITICAL FIX: Use SFTConfig directly to bypass Unsloth's broken translation layer
    args = SFTConfig(
        per_device_train_batch_size = 4,
        gradient_accumulation_steps = 2,
        warmup_steps = 25,                 # Increased for stable initialization
        num_train_epochs = 3,              # Reduced to prevent overfitting
        learning_rate = 5e-5,              # Lowered for 26B parameter scale
        fp16 = not is_bfloat16,
        bf16 = is_bfloat16,
        logging_steps = 1,
        optim = "adamw_8bit",
        weight_decay = 0.01,
        lr_scheduler_type = "cosine",      # Changed for better convergence
        seed = 3407,
        output_dir = os.path.join(project_dir, "outputs"),
        dataset_text_field = "text", # Moved inside SFTConfig as required by newer TRL
        max_seq_length = 8192,       # Moved inside SFTConfig as required by newer TRL
    ),
)

trainer.train()
```

### Fine-Tuned Model Evaluation (After Training)
Now let's run the exact same prompt through the model to see if it successfully adopted the structured `<|think|>` deliberation format and source verification.

```python
FastModel.for_inference(model) # Enable native 2x faster inference

# 1. Use explicit 'text=' keyword to bypass the 'NoneType' image bug
# 2. Prevent double-BOS tokens if your prompt string already has one
inputs = tokenizer(
    text = [prompt],
    return_tensors = "pt",
    add_special_tokens = False
).to("cuda")

# Increased to 1024 to accommodate complex hypothesis enumeration
outputs = model.generate(**inputs, max_new_tokens = 1024, use_cache = True)
# CRITICAL FIX: skip_special_tokens must be False to see your custom tokens!
tuned_response = tokenizer.batch_decode(outputs, skip_special_tokens = False)[0]

print("========== FINE-TUNED MODEL RESPONSE ==========")
print(tuned_response)
print("=========================================")

def evaluate_structure(response):
    metrics = {
        "Has Logical Chain": "<unused5>" in response or "[Logical Chain]" in response,
        "Has Competing Hypotheses": "<unused6>" in response or "[Competing Hypotheses]" in response,
        "Has Discarded Paths": "<unused7>" in response or "[Discarded Paths]" in response or "✗ Discarded:" in response,
        "Has Tool Call": "<unused3>verify_claim" in response and "<unused4>" in response,
        "Has Final Output Signal": "🟢" in response or "🟡" in response or "🔴" in response,
        "Has Confidence Badge": "Confidence:" in response
    }
    print("\n--- Structured Output Evaluation ---")
    score = sum(1 for passed in metrics.values() if passed)
    for key, passed in metrics.items():
        print(f"{key}: {'✅ Pass' if passed else '❌ Fail'}")
    print(f"\nOverall Structure Score: {score}/{len(metrics)}")

print("\nEvaluating Base Model:")
evaluate_structure(base_response)

print("\nEvaluating Fine-Tuned Model:")
evaluate_structure(tuned_response)
```

```python
def run_inference(eval_prompt, max_tokens=3072):

    prompt_str = f"{strict_system_prompt}\n<bos><start_of_turn>user\n{eval_prompt}<end_of_turn>\n<start_of_turn>model\n<unused0>\n"
    prompt_tokens = llm.tokenize(prompt_str.encode("utf-8"), add_bos=False, special=True)

    output = llm(
        prompt_tokens,
        max_tokens=max_tokens,
        temperature=0.3,
        top_p=0.85,
        stop=["<eos>", "<end_of_turn>"],
        grammar=grammar,
        echo=False
    )
    return output['choices'][0]['text']
def evaluate_structure(response, expected_dot=None):
    metrics = {
        "has_logical_chain": "<unused5>" in response or "[Logical Chain]" in response,
        "has_hypotheses": "<unused6>" in response or "[Competing Hypotheses]" in response,
        "has_discarded_paths": "<unused7>" in response or "[Discarded Paths]" in response or "✗ Discarded:" in response,
        "has_signal_dot": "🟢" in response or "🟡" in response or "🔴" in response,
        "has_confidence_badge": "Confidence:" in response
    }

    # Calculate base score out of 5 mandatory structural elements
    metrics["score"] = sum(1 for passed in metrics.values() if passed)
    metrics["total"] = 5

    # Log conditional elements separately
    metrics["has_tool_call"] = "<unused3>verify_claim" in response and "<unused4>" in response

    if expected_dot:
        metrics["matches_expected_dot"] = expected_dot in response

    return metrics

def print_eval(metrics):
    print(f"Structure Score: {metrics['score']}/{metrics['total']}")
    for k, v in metrics.items():
        if k not in ["score", "total", "query"]:
            print(f"  {k}: {'✅' if v else '❌'}")

HOLDOUT_TESTS = [
    # Severe interactions (should produce 🔴)
    {"prompt": "Patient on Simvastatin 80mg just started Itraconazole 200mg for toenail fungus. Evaluate.", "expected_dot": "🔴"},

    # Moderate interactions (should produce 🟡)
    {"prompt": "72yo female on Digoxin 0.25mg and Amiodarone 200mg. Review all interactions.",
     "expected_dot": "🟡"},

    # Safe combinations (should produce 🟢)
    {"prompt": "Patient on Metformin 500mg and Atorvastatin 20mg. Any interaction?",
     "expected_dot": "🟢"},

    # Complex polypharmacy (tests full deliberation depth)
    {"prompt": (
        "82yo nursing home resident on: Warfarin 5mg, Omeprazole 40mg, "
        "Clopidogrel 75mg, Aspirin 81mg, Metoprolol 50mg, Sertraline 100mg, "
        "Gabapentin 300mg, Amlodipine 10mg, Metformin 1000mg, Glipizide 10mg. "
        "Identify the three most dangerous interactions."
    ), "expected_dot": "🔴"}
]

print(f"Running {len(HOLDOUT_TESTS)} holdout tests...\n")
all_metrics = []

for i, test in enumerate(HOLDOUT_TESTS, 1):
    prompt = test["prompt"]
    expected = test["expected_dot"]

    print(f"{'='*60}")
    print(f"TEST {i}/{len(HOLDOUT_TESTS)}")
    print(f"QUERY: {prompt[:100]}...")
    print(f"EXPECTED: {expected}")
    print(f"{'='*60}")

    response = run_inference(prompt)
    metrics = evaluate_structure(response, expected_dot=expected)
    metrics["query"] = prompt[:80]
    all_metrics.append(metrics)
    print_eval(metrics)
    print()
```

```python
import statistics

# Must strictly match the keys returned by evaluate_structure
fields = [
    "has_logical_chain", "has_hypotheses", "has_discarded_paths",
    "has_tool_call", "has_signal_dot", "has_confidence_badge",
    "matches_expected_dot"
]

print("=" * 60)
print("P.R.I.S.M. STRUCTURAL COMPLIANCE SUMMARY")
print("=" * 60)

for field in fields:
    hits = sum(1 for m in all_metrics if m.get(field, False))
    pct = (hits / len(all_metrics)) * 100 if all_metrics else 0
    bar = "█" * int(pct / 5) + "░" * (20 - int(pct / 5))
    label = field.replace("has_", "").replace("_", " ").title()
    print(f"{label:<25} {bar} {hits}/{len(all_metrics)} ({pct:.0f}%)")

scores = [m["score"] for m in all_metrics]
avg = statistics.mean(scores) if scores else 0
total = all_metrics[0]["total"] if all_metrics else 1
overall = (avg / total) * 100

pass_threshold = int(total * 0.75)

print(f"\n{'─' * 60}")
print(f"  Overall Compliance: {overall:.1f}% ({avg:.1f}/{total} avg)")
print(f"  Tests Passed (≥{pass_threshold}/{total}): {sum(1 for s in scores if s >= pass_threshold)}/{len(scores)}")
print(f"{'─' * 60}")
```

```python
import re

def display_glass_box(response: str):
    """Parse and pretty-print a P.R.I.S.M. response."""
    thought = ""
    answer = ""

    # 1. Primary Parsing: Split on the channel router mapping to <unused2>
    parts = response.split("<unused2>")

    if len(parts) >= 2:
        thought_part = parts[0]
        if "<unused1>" in thought_part:
            thought = thought_part.split("<unused1>")[-1].strip()
        else:
            thought = thought_part.replace("<unused0>", "").strip()
        answer = parts[1].strip()
    # 2. Robust Fallback: In case the model fails to output the closing channel tag
    else:
        match = re.search(r'(🟢|🟡|🔴)', response)
        if match:
            split_idx = match.start()
            thought = response[:split_idx].replace("<unused0>", "").replace("<unused1>", "").strip()
            answer = response[split_idx:].strip()
        else:
            thought = response.replace("<unused0>", "").replace("<unused1>", "").strip()
            answer = "(Model failed to utilize channel tags or signal dots)"

    # Restore human-readable tags for display
    thought = thought.replace("<unused5>", "[Logical Chain]").replace("<unused6>", "[Competing Hypotheses]").replace("<unused7>", "[Discarded Paths]").replace("<unused8>", "▶ Selected:")
    answer = answer.replace("<unused3>", "<|tool_call>").replace("<unused4>", '<|"|>')

    # Print Formatting
    print("┌─── 🧠 DELIBERATION TRACE (Expert View) ─────────────┐")
    for line in thought.split("\n"):
        if line.strip():
            print(f"│  {line.strip()}")
    print("└─────────────────────────────────────────────────────┘\n")

    print("┌─── 📋 CLINICAL ANSWER (Default View) ───────────────┐")
    for line in answer.split("\n"):
        if line.strip():
            print(f"│  {line.strip()}")
    print("└─────────────────────────────────────────────────────┘")
```
